# Obtención de datos geográficos — INEGI

## Proyecto Agronomía

Esta libreta documenta la obtención y preparación de la información
geográfica utilizada en el proyecto.

Las geometrías de las entidades federativas y municipios de México
se obtienen mediante los servicios geográficos de INEGI.

La lógica de adquisición y procesamiento se encuentra implementada
en el paquete `proyecto_agronomia`, por lo que esta libreta funciona
como una interfaz reproducible para ejecutar, inspeccionar y validar
el proceso.

## Objetivo

Obtener:

- geometrías de los 32 estados de México;
- geometrías municipales;
- coordenadas representativas de cada municipio.

Estas coordenadas serán utilizadas posteriormente para consultar
las variables climáticas históricas.

## Flujo de trabajo

1. Descargar o cargar las geometrías municipales de INEGI.
2. Descargar o cargar las geometrías estatales.
3. Generar un punto representativo para cada municipio.
4. Validar claves geográficas y coordenadas.
5. Guardar el resultado para la etapa climática.

```python
from proyecto_agronomia.dataset import (
    obtener_estados_inegi,
    obtener_municipios_inegi,
)

from proyecto_agronomia.features import (
    guardar_coordenadas_municipios,
)
```

## 1. Geometrías municipales

Se obtienen las geometrías de los municipios de México.

La función reutiliza el archivo local cuando este ya existe, evitando
realizar nuevamente la descarga.

```python
municipios_geo = obtener_municipios_inegi()

municipios_geo.head()
```

```python
print(f"Municipios disponibles: {len(municipios_geo):,}")
print(f"CRS: {municipios_geo.crs}")
print(f"Columnas: {municipios_geo.columns.tolist()}")
```

## 2. Geometrías estatales

Se obtienen las geometrías correspondientes a las 32 entidades
federativas de México.

```python
estados_geo = obtener_estados_inegi()

estados_geo.head()
```

```python
print(f"Entidades federativas: {len(estados_geo):,}")
print(f"CRS: {estados_geo.crs}")
```

## 3. Coordenadas representativas municipales

Para consultar posteriormente información climática es necesario
asociar cada municipio con una coordenada.

Se calcula un punto representativo dentro de cada geometría municipal
y posteriormente se transforma al sistema de coordenadas geográficas
EPSG:4326.

```python
municipios_coordenadas = guardar_coordenadas_municipios(
    municipios_geo
)

municipios_coordenadas.head()
```

```python
print(
    f"Municipios: {len(municipios_coordenadas):,}"
)

print(
    "Claves duplicadas:",
    municipios_coordenadas.duplicated(
        ["Idestado", "Idmunicipio"]
    ).sum(),
)

print(
    "Coordenadas faltantes:",
    municipios_coordenadas[
        ["Latitud", "Longitud"]
    ].isna().sum().sum(),
)

print(
    f"Latitud: "
    f"{municipios_coordenadas['Latitud'].min():.4f} → "
    f"{municipios_coordenadas['Latitud'].max():.4f}"
)

print(
    f"Longitud: "
    f"{municipios_coordenadas['Longitud'].min():.4f} → "
    f"{municipios_coordenadas['Longitud'].max():.4f}"
)
```

## Resultado

La información geográfica de INEGI quedó preparada para las siguientes
etapas del proyecto.

El conjunto de coordenadas municipales utiliza como identificadores:

- `Idestado`
- `Idmunicipio`

Estas claves permiten relacionar posteriormente la información
geográfica con los registros agrícolas.

Las coordenadas `Latitud` y `Longitud` se utilizarán para consultar
los datos climáticos históricos de cada municipio.
