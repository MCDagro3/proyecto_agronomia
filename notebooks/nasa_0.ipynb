# Clima desde la NASA.

```python
from yarl import URL
import requests


nasa = URL("https://power.larc.nasa.gov/api/temporal/daily/point")

start = "20030101"
end = "20040101"
laritude, longitude = -10, 10
community = "ag"
parameters = "T2M"
header = "true"
units = "metric"
format = "json"

api_parameters = {
    "start": start,
    "end": end,
    "latitude": laritude,
    "longitude": longitude,
    "community": community,
    "parameters": parameters,
    "format": format
    }

call = nasa.with_query(api_parameters)
response = requests.get(call)
response.json()
```
