#### Variables Categoricas
En papel de momento. Pero a eso iremos.

```python
import re


def long_rule(val, long):
    assert str(val), "Revisa la entrada"
    return len(str(val)) <= long

def range_rule(val, rage: tuple):
    a, b = rage
    check_set = set(range(a, b+1))
    return int(val) in check_set


# Reglas de Variables categoticas
## Caso particular 
rl_mun = rules.loc['Idmunicipio']
long = rl_mun['Long.']
range_mun = rl_mun["Rangos\nválidos"]
datarules = {"Idmunicipio": lambda x, l=long: long_rule(x, l)}


cat_vars = rules[rules['Tipo']=="N"]
cat_vars.drop(axis=0, index="Anio", inplace=True)
for var in cat_vars.index:
    k = re.search(r"{(.*)}",cat_vars.loc[var, "Rangos\nválidos"])
    ranges = k.group(1)
    g_min = ranges.split(",")[0]
    g_max = ranges.split(",")[-1]
    datarules[var] = lambda x, ival=(int(g_min), int(g_max)): range_rule(x, ival)

```

#### Variables de Texto y Variables Numéricas

```python
text_vars = rules[rules['Tipo']=="T"]
text_vars.drop(axis=0, index="Idmunicipio", inplace=True)
text_vars
```

```python
# Reglas para variables de texto
for var in text_vars.index:
    # print(var,text_vars.loc[var, "Long."])
    datarules[var] = lambda x, l = text_vars.loc[var, "Long."]: long_rule(x, l)


# Reglas para variables numéricas.
def interval_rule(val, interval: tuple, config=(True, True)):
    val = float(val)
    try:
        a, b = interval
        if config[0] :
            if config[1]:
                return a <= val <= b
            else:
                return a <= val < b
        else:
            if config[1]:
                return a < val <= b
            else:
                return a < val < b
    except:
        print(val)
        print(f"SI {val} es un valor nulo. Debe revisar el contexto")
        return 1


# Caso particular. El año.
rl_year = rules.loc['Anio']
nd = rl_year['Long.']
my = 0
mmy = 10 ** nd
datarules['Anio'] = lambda x, ival=(my, mmy): interval_rule(x, ival, config=(True, False))


num_vars = rules[rules['Tipo']=="D"]

for var in num_vars.index:
    k = re.search(r"{(.*)}",num_vars.loc[var, "Rangos\nválidos"])
    ranges = k.group(1)
    dec = len(ranges.split(".")[-1])
    nd = num_vars.loc[var, "Long."]
    m = 0
    mm = 10 ** nd - 1
    datarules[var] = lambda x, ival=(m, mm): interval_rule(x, ival)
    

print(datarules)
```

Con eso ya tenemos un conjunto de reglas para cada columna de nuestro conjunto de datos. Ahora revisamos si las columnas satifacen las reglas previamente establecidas.

```python
# print(prod_agricola.head(n=5))

bad_columns = []

for col in prod_agricola.columns:
    check_rule = datarules[col]
    validate = prod_agricola[col].apply(check_rule)
    if not validate.all():
        bad_columns.append(col)
        print(f"Error en la columna {col}")
```

Vemos que hay 5 columnas que por lo menos poseen un valor que no satisface las reglas establecidas por el diccionario. Revisando las columnas podemos ver los elementos que no cumplen las reglas.

```python
col = bad_columns[0]

# El nombre del municipio no debe exceder los 45 caracteres.
print(rules.loc[col])

# Filtrar las filas que no cumplen con la regla
fail_col = prod_agricola[col]
fail_col = fail_col[~fail_col.apply(datarules[col])]
fail_df = pd.DataFrame(fail_col)
fail_df['Long'] = fail_col.apply(lambda x: len(str(x)))
fail_rate = len(fail_df) / len(prod_agricola) * 100
print("Valores fuera de regla", len(fail_df))
fail_rate
```

#### Variante usando pandera.

Ahora que ya vimos una idea general de como se hace. Para el proyecto lo haremos usando Pandera y su complemento para pandas.

```python
import pandera.pandas as pa

agro_schema = pa.DataFrameSchema({
    col: pa.Column(prod_agricola[col].dtype, pa.Check(datarules[col]), nullable=True)
    },
    index = pa.Index(int),
    name="dgsiap_mun"
)

try:
    agro_schema.validate(prod_agricola, lazy=True)
except pa.errors.SchemaErrors as exc:
    print(exc)
```
