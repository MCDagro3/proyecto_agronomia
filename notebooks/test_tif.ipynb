```python
# Opción 1: rasterio (estándar, basado en GDAL — combina bien con geopandas)
import rasterio
with rasterio.open("/home/dusty/Documents/Repositories/proyecto_agronomia/preload/28464494/norm9120precipyear.tif") as src:
    print(src.meta)          # dimensiones, dtype, crs, nodata...
    print(src.crs)           # sistema de coordenadas
    print(src.bounds)        # extensión geográfica
    precip = src.read(1)     # numpy array (alto, ancho) con mm de lluvia
    for fila in src.sample([(-99.13, 19.43)]):  # CDMX, si el CRS es WGS84
        print("Precipitación:", fila[0], "mm")


import numpy as np
import pandas as pd
with rasterio.open("/home/dusty/Documents/Repositories/proyecto_agronomia/preload/28464494/norm9120precipyear.tif") as src:
    filas, cols = np.where(np.isfinite(precip))
    print(filas, cols)
    xs, ys = rasterio.transform.xy(src.transform, filas, cols)
    df = pd.DataFrame({
        "lon": xs, "lat": ys,
        "precip": precip[filas, cols],
    })
```
